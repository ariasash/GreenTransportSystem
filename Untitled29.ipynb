```python
import speech_recognition as sr
from googletrans import Translator
from gtts import gTTS
from IPython.display import Audio, display, HTML
from google.colab import files
import io
import os
from pydub import AudioSegment
from google.colab import output
import base64

print("✅ Librerías cargadas correctamente.")
```

```python
# --- PARTE DE CONFIGURACIÓN ---
# Nombre fijo para el archivo de salida
OUTPUT_MP3_FILENAME = "grabacion_colab.mp3"

# --- Parte de JavaScript/HTML (CAMBIADO a audio/webm) ---
RECORDER_HTML = """
<script>
let mediaRecorder;
let audioChunks = [];
let stream;

async function startRecording() {
  try {
    stream = await navigator.mediaDevices.getUserMedia({ audio: true });
    mediaRecorder = new MediaRecorder(stream);
    audioChunks = [];

    mediaRecorder.ondataavailable = (event) => {
      audioChunks.push(event.data);
    };

    mediaRecorder.onstop = async () => {
      const audioBlob = new Blob(audioChunks, { type: 'audio/webm' });
      const reader = new FileReader();
      reader.readAsDataURL(audioBlob);
      reader.onloadend = () => {
        const base64Audio = reader.result.split(',')[1];
        google.colab.kernel.invokeFunction('notebook.receive_audio', [base64Audio], {});
      };
    };

    mediaRecorder.start();
    document.getElementById('status').innerHTML = '🔴 Grabando... (haz clic en Detener)';
    document.getElementById('startBtn').disabled = true;
    document.getElementById('stopBtn').disabled = false;
  } catch(error) {
    document.getElementById('status').innerHTML = '❌ Error al acceder al micrófono. Verifica los permisos.';
    console.error('Error al iniciar la grabación:', error);
  }
}

function stopRecording() {
  if (mediaRecorder && mediaRecorder.state !== 'inactive') {
    mediaRecorder.stop();
    stream.getTracks().forEach(track => track.stop());
    document.getElementById('status').innerHTML = '⏳ Procesando audio...';
    document.getElementById('startBtn').disabled = false;
    document.getElementById('stopBtn').disabled = true;
  }
}
</script>

<div style="padding: 20px; background: #f0f0f0; border-radius: 10px; text-align: center; font-family: sans-serif;">
  <h3>🎤 Grabador de Voz</h3>
  <button id="startBtn" onclick="startRecording()"
          style="padding: 15px 30px; font-size: 16px; background: #4CAF50; color: white; border: none; border-radius: 5px; cursor: pointer; margin: 5px;">
    ▶️ Iniciar Grabación
  </button>
  <button id="stopBtn" onclick="stopRecording()" disabled
          style="padding: 15px 30px; font-size: 16px; background: #f44336; color: white; border: none; border-radius: 5px; cursor: pointer; margin: 5px;">
    ⏹️ Detener
  </button>
  <p id="status" style="margin-top: 15px; font-size: 14px;">Listo para grabar</p>
  <p id="result" style="margin-top: 10px; font-size: 14px; font-weight: bold;"></p>
</div>
"""

# --- Parte de Python (Callback y Conversión) ---
def receive_audio(base64_data):
    # Decodificar y convertir a MP3 usando el nombre de archivo fijo
    try:
        audio_data_webm = base64.b64decode(base64_data)
        audio_buffer = io.BytesIO(audio_data_webm)

        # Cargar WebM y Exportar a MP3
        audio = AudioSegment.from_file(audio_buffer, format="webm")
        audio.export(OUTPUT_MP3_FILENAME, format="mp3", bitrate="128k")

        # Informar al usuario
        size_bytes = os.path.getsize(OUTPUT_MP3_FILENAME)
        size_mb = size_bytes / (1024 * 1024)

        print(f"\n🎵 Audio convertido y guardado como: **{OUTPUT_MP3_FILENAME}** ({size_mb:.2f} MB)")

        js_update_result = f"""
        document.getElementById('status').innerHTML = '✅ Audio Guardado';
        document.getElementById('result').innerHTML = 'Archivo: {OUTPUT_MP3_FILENAME} ({size_mb:.2f} MB)';
        """
        display(HTML(f'<script>{js_update_result}</script>'))

    except Exception as e:
        print(f"❌ Error durante la conversión a MP3: {e}")
        js_update_error = f"""
        document.getElementById('status').innerHTML = '❌ Error de conversión';
        document.getElementById('result').innerHTML = 'Verifica la consola de Python';
        """
        display(HTML(f'<script>{js_update_error}</script>'))


# Registrar la función de callback
output.register_callback('notebook.receive_audio', receive_audio)

print("--- INTERFAZ DEL GRABADOR ---")
display(HTML(RECORDER_HTML))
```

```python
# Reemplaza la función transcribir_audio en la Celda 4 con esto:
# ------------------------------
# 1. FUNCIÓN DE RECONOCIMIENTO DE VOZ (STT) - VERSIÓN ROBUSTA
# ------------------------------
def transcribir_audio(audio_filepath, idioma_origen='es-ES'):
    """
    Convierte audio a texto. Utiliza pydub para convertir temporalmente
    el MP3 a WAV, asegurando la compatibilidad con SpeechRecognition.
    """
    recognizer = sr.Recognizer()
    TEMP_WAV_FILE = "temp_colab.wav"

    if not os.path.exists(audio_filepath):
        print(f"❌ Error: El archivo de audio '{audio_filepath}' no se encontró.")
        return None

    try:
        # 1. Usar pydub para cargar el MP3
        print("   -> Convirtiendo MP3 a WAV temporal para compatibilidad...")
        # NOTA: Debes importar AudioSegment de pydub en la Celda 2/3
        audio_mp3 = AudioSegment.from_mp3(audio_filepath)

        # 2. Exportar a WAV (formato preferido por SpeechRecognition)
        audio_mp3.export(TEMP_WAV_FILE, format="wav")

        # 3. Usar SpeechRecognition con el archivo WAV temporal
        with sr.AudioFile(TEMP_WAV_FILE) as source:
            recognizer.adjust_for_ambient_noise(source, duration=0.5)
            audio = recognizer.record(source)

        # 4. Usar Google Speech Recognition (API gratuita)
        texto = recognizer.recognize_google(audio, language=idioma_origen)
        print(f"📝 Texto original ({idioma_origen}): {texto}")
        return texto

    except sr.UnknownValueError:
        print("❌ No se pudo entender el audio. Intenta hablar más claro o revisa el idioma.")
        return None

    except Exception as e:
        print(f"❌ Error durante la transcripción: {e}")
        return None

    finally:
        # Limpiar el archivo WAV temporal después de usarlo
        if os.path.exists(TEMP_WAV_FILE):
            os.remove(TEMP_WAV_FILE)

# ... (El resto de las funciones de la Celda 4 deben seguir abajo)
# ------------------------------
# 2. FUNCIÓN DE TRADUCCIÓN (MT)
# ------------------------------
def traducir_texto(texto, idioma_destino='en'):
    """
    Traduce texto a otro idioma usando Google Translate (googletrans).
    """
    translator = Translator()

    try:
        # Detectar idioma origen automáticamente y traducir
        traduccion = translator.translate(texto, dest=idioma_destino)

        print(f"🌍 Idioma detectado: {traduccion.src}")
        print(f"🔄 Traducción ({idioma_destino}): {traduccion.text}")

        return traduccion.text, traduccion.src
    except Exception as e:
        print(f"❌ Error en traducción: {e}")
        return None, None

# ------------------------------
# 3. FUNCIÓN DE SÍNTESIS DE VOZ (TTS)
# ------------------------------
def generar_audio_mp3(texto, idioma='en', nombre_archivo='traduccion.mp3'):
    """
    Convierte texto a audio MP3 usando Google Text-to-Speech (gTTS).
    """
    try:
        tts = gTTS(text=texto, lang=idioma, slow=False)
        tts.save(nombre_archivo)
        print(f"🔊 Audio generado: {nombre_archivo}")
        return nombre_archivo
    except Exception as e:
        print(f"❌ Error generando audio: {e}")
        return None
```

```python
OUTPUT_MP3_FILENAME = "grabacion_colab.mp3" # Nombre del archivo grabado en la Celda 3
OUTPUT_TRANSLATION_FILENAME = "traduccion.mp3"

def traductor_completo(idioma_origen='es', idioma_destino='en'):
    """
    Función principal que orquesta todo el proceso.
    """

    # ⚠️ REVISIÓN CRUCIAL: Verificar si el archivo MP3 existe después de grabar.
    if not os.path.exists(OUTPUT_MP3_FILENAME):
        print("="*50)
        print("⚠️ Primero, ejecuta la Celda 3, presiona 'Iniciar Grabación', habla y haz clic en 'Detener'.")
        print("="*50)
        return

    print("="*50)
    print("🚀 INICIANDO PROCESO DE TRADUCCIÓN")
    print("="*50)

    # Paso 1: Transcribir audio a texto (Speech-to-Text)
    print("\n📌 PASO 1: Reconocimiento de voz (Speech-to-Text)")
    # El código para STT debe ser de 5 letras (ej: es-ES)
    idioma_stt = f'{idioma_origen}-{idioma_origen.upper()}'
    texto_original = transcribir_audio(OUTPUT_MP3_FILENAME, idioma_stt)

    if not texto_original:
        return

    # Paso 2: Traducir texto (Machine Translation - NLP)
    print("\n📌 PASO 2: Traducción (NLP)")
    texto_traducido, idioma_detectado = traducir_texto(texto_original, idioma_destino)

    if not texto_traducido:
        return

    # Paso 3: Generar audio de la traducción (Text-to-Speech)
    print("\n📌 PASO 3: Síntesis de voz (Text-to-Speech)")
    archivo_audio = generar_audio_mp3(texto_traducido, idioma_destino, OUTPUT_TRANSLATION_FILENAME)

    # Paso 4: Mostrar resultados en pantalla
    print("\n" + "="*50)
    print("📊 RESULTADOS FINALES")
    print("="*50)

    # Crear tabla de resultados (Mostrando en pantalla con HTML)
    resultado_html = f"""
    <div style="padding: 20px; background: linear-gradient(135deg, #667eea 0%, #764ba2 100%);
                  border-radius: 15px; color: white; margin: 10px 0;">
        <h2 style="margin: 0 0 15px 0;">🎯 Traducción Completada</h2>

        <div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 10px; margin: 10px 0;">
            <strong>🗣️ Texto Original ({idioma_origen.upper()}):</strong>
            <p style="font-size: 18px; margin: 5px 0;">{texto_original}</p>
        </div>

        <div style="background: rgba(255,255,255,0.2); padding: 15px; border-radius: 10px; margin: 10px 0;">
            <strong>🌍 Traducción ({idioma_destino.upper()}):</strong>
            <p style="font-size: 18px; margin: 5px 0;">{texto_traducido}</p>
        </div>
    </div>
    """
    display(HTML(resultado_html))

    # Reproducir audio
    if archivo_audio and os.path.exists(archivo_audio):
        print("\n🔊 Audio de la traducción:")
        display(Audio(archivo_audio, autoplay=True))

        # Opción para descargar
        print("\n📥 Haz clic para descargar el MP3:")
        files.download(archivo_audio)

    return texto_original, texto_traducido, archivo_audio
```
